```python
# Check if the dataset contains the values "UNKNOWN" or "ERROR"
# This creates a boolean DataFrame where True indicates cells containing either "UNKNOWN" or "ERROR"
# The .any() method returns True for each column that has at least one occurrence of these values
df.isin(["UNKNOWN", "ERROR"]).any()
```

```python
# Identify columns that contain "UNKNOWN" or "ERROR"
# This line finds all columns in the dataframe where at least one cell contains either "UNKNOWN" or "ERROR"
# df.isin() creates a boolean mask where True indicates the presence of the specified values
# .any() checks if any value in each column matches the condition
# The result is a Series of column names that meet this criteria
df.columns[df.isin(["UNKNOWN", "ERROR"]).any()]
```

```python
# Count occurrences of "UNKNOWN" and "ERROR" in the dataset
# This creates a boolean mask where True indicates cells containing either "UNKNOWN" or "ERROR"
# Then sum() counts the True values for each column, showing how many times these values appear
df.isin(["UNKNOWN", "ERROR"]).sum()
```

```python
# The dataset contains some incorrect values such as "UNKNOWN" or "ERROR".
# These values indicate missing, invalid, or unreliable data that should not be used for analysis.

# df.isin(["UNKNOWN", "ERROR"]) checks the entire DataFrame and marks the cells where these values appear.

# .any(axis=1) checks each row and returns True if at least one column in that row contains "UNKNOWN" or "ERROR".

# The ~ symbol means NOT. It reverses the condition so that rows containing these invalid values are excluded.

# As a result, only rows with valid and clean data are kept in the DataFrame.

df = df[~df.isin(["UNKNOWN", "ERROR"]).any(axis=1)]
```

```python
# Display basic information about the dataset such as
# number of rows, columns, and data types.

df.info()
```

```python
# Count missing values in each column.
# Missing values can cause problems during analysis.

df.isnull().sum()
```

```python
# Check for missing values in the dataframe and remove rows with any missing values
# The inplace=True parameter modifies the original dataframe without needing reassignment
df.dropna(inplace = True)
```

```python
df.isnull().sum()
```

```python
# Check if duplicate rows exist in the dataset.

df.duplicated().sum()
```

```python
# Remove duplicate records to maintain data integrity.

df = df.drop_duplicates()
```

```python
# Identify how many invalid values such as "UNKNOWN" or "ERROR"
# exist in each column.

df.isin(["UNKNOWN", "ERROR"]).sum()
```

```python
# Remove rows that contain invalid values such as "UNKNOWN" or "ERROR".
# These values indicate incorrect or unreliable data.

df = df[~df.isin(["UNKNOWN", "ERROR"]).any(axis=1)]
```

```python
df.dtypes
```

```python
# Convert Transaction ID to string format because it contains text identifiers.

df["Transaction ID"] = df["Transaction ID"].astype("string")
```

```python
# Convert Item column to string since it stores item names.

df["Item"] = df["Item"].astype("string")
```

```python
# Convert Quantity to integer because it represents the number of items purchased.

df["Quantity"] = df["Quantity"].astype(int)
```

```python
# Convert Price Per Unit to float to allow decimal values for prices.

df["Price Per Unit"] = df["Price Per Unit"].astype(float)
```

```python
# Convert Total Spent to float since it represents monetary values.

df["Total Spent"] = df["Total Spent"].astype(float)
```

```python
# Convert Payment Method to string since it contains text values
# such as Cash, Credit Card, or Digital Wallet.

df["Payment Method"] = df["Payment Method"].astype("string")
```

```python
# Convert Location to string because it stores descriptive text
# such as In-store or Takeaway.

df["Location"] = df["Location"].astype("string")
```

```python
# Convert Transaction Date from text format to datetime format.
# This allows proper date-based analysis such as filtering by month or year.

df["Transaction Date"] = pd.to_datetime(df["Transaction Date"])
```

```python
# Save the cleaned dataset to a new CSV file.
df.to_csv("cleaned_sales_data.csv", index=False)
```

```python
# Validate whether the Total Spent value matches the calculation
# Quantity multiplied by Price Per Unit.
# This helps identify incorrect or inconsistent financial records
# that may have occurred due to data entry errors.
df["Calculated Total"] = df["Quantity"] * df["Price Per Unit"]
```

```python
# Identify rows where the stored Total Spent does not match the calculated value
invalid_totals = df[df["Total Spent"] != df["Calculated Total"]]
```

```python
# Display rows where the values do not match
invalid_totals
```

```python
# Check whether each Transaction ID follows the expected format
# where it begins with the prefix "TXN_".
# This ensures that all transaction identifiers follow a consistent structure.
df["Valid Transaction ID"] = df["Transaction ID"].str.startswith("TXN_")
```

```python
# Count how many IDs are valid and invalid
df["Valid Transaction ID"].value_counts()
```
